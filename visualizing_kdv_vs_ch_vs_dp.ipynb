```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-6, 6, 1000)

# kdv soliton, c=2
c = 2.0
kdv = (c / 2) * (1 / np.cosh((np.sqrt(c) / 2) * x))**2

# peakon, c=1
peakon = np.exp(-np.abs(x))

# shockpeakon (dp), k=0 snapshot at t=1
shockpeakon = -(1 / (1)) * np.sign(x) * np.exp(-np.abs(x))

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].plot(x, kdv, color='steelblue', linewidth=2)
axes[0].set_title('KdV soliton', fontsize=13)
axes[0].set_xlabel('x', fontsize=11)
axes[0].set_ylabel('u', fontsize=11)
axes[0].set_ylim(-0.2, 1.4)
axes[0].axhline(0, color='gray', linewidth=0.5, linestyle='--')
axes[0].grid(True, alpha=0.3)

axes[1].plot(x, peakon, color='darkorange', linewidth=2)
axes[1].set_title('Peakon (CH / DP)', fontsize=13)
axes[1].set_xlabel('x', fontsize=11)
axes[1].set_ylim(-0.2, 1.4)
axes[1].axhline(0, color='gray', linewidth=0.5, linestyle='--')
axes[1].grid(True, alpha=0.3)

axes[2].plot(x, shockpeakon, color='crimson', linewidth=2)
axes[2].set_title('Shockpeakon (DP)', fontsize=13)
axes[2].set_xlabel('x', fontsize=11)
axes[2].set_ylim(-1.4, 1.4)
axes[2].axhline(0, color='gray', linewidth=0.5, linestyle='--')
axes[2].grid(True, alpha=0.3)

fig.suptitle('Wave profiles: smooth to non-smooth', fontsize=14, y=1.02)
plt.tight_layout()
plt.savefig('fig11_profiles.png', dpi=150, bbox_inches='tight')
print('saved fig11_profiles.png')
```
